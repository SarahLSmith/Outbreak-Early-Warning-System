```python
#Define storage account
storage_account_name = "ukhsadev2026"

#Define credintials (public "nametags" to identify accounts)
client_id ="5c9267a6-7683-40d8-bbf2-a5512db4c402"
tenant_id = "8c7eb586-568c-41d2-9fb6-00e5d3695d80"

#Retriving the client secret
client_secret = dbutils.secrets.get(scope="ukhsa-pipeline-scope", key="adls-client-secret")

#Configure Spark to login to Azure Data Lake Storage Gen2
#Login to Azure using oauth
spark.conf.set(f"fs.azure.account.auth.type.{storage_account_name}.dfs.core.windows.net", "OAuth")
#Specifying through which provider. (Login as a robot)
spark.conf.set(f"fs.azure.account.oauth.provider.type.{storage_account_name}.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider")
#Set the username
spark.conf.set(f"fs.azure.account.oauth2.client.id.{storage_account_name}.dfs.core.windows.net", client_id)
#Set the password
spark.conf.set(f"fs.azure.account.oauth2.client.secret.{storage_account_name}.dfs.core.windows.net", client_secret)
#Login to Azure
spark.conf.set(f"fs.azure.account.oauth2.client.endpoint.{storage_account_name}.dfs.core.windows.net", f"https://login.microsoftonline.com/{tenant_id}/oauth2/token")

#Test connection to bronze container
display(dbutils.fs.ls(f"abfss://bronze@{storage_account_name}.dfs.core.windows.net/"))
```
